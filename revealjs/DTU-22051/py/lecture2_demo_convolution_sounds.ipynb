# Lecture 2 - DEMO
Recording an impulse response and convolving it with a sound signal.

```python
import numpy as np
import sounddevice as sd
import soundfile as sf
from scipy.signal import convolve
from IPython.display import Audio
```

```python
print(sd.query_devices())
```

```python
fs = 44100
nbits = 16
rec_id = 3
play_id = 2
path2audio = "../sounds/"
audio_fname = "lecture01_demo_mini-me_short.wav"
```

```python
duration = 2
print("Recording impulse response...")
h = sd.rec(int(duration*fs), samplerate=fs, channels=1, dtype="float32", device=rec_id)
sd.wait()
h = h.flatten()
```

```python
sd.play(h, fs, device=play_id)
sd.wait()
Audio(h, rate=fs)
```

```python
duration = 5
print("Recording signal...")
sig = sd.rec(int(duration*fs), samplerate=fs, channels=1, dtype="float32", device=rec_id)
sd.wait()
sig = sig.flatten()
```

```python
sd.play(sig, fs, device=play_id)
sd.wait()
Audio(sig, rate=fs)
```

```python
sig_h = convolve(h, sig)
sd.play(sig_h, fs, device=play_id)
sd.wait()
Audio(sig_h, rate=fs)
```

```python
sig_wav, fs_wav = sf.read(path2audio + audio_fname)
if sig_wav.ndim > 1:
    sig_wav = np.mean(sig_wav, axis=1)
print(fs_wav)
```

```python
sig_wav_h = convolve(h, sig_wav)
sig_wav_h = sig_wav_h / np.max(np.abs(sig_wav_h)) * 0.8
sd.play(sig_wav_h, fs, device=play_id)
sd.wait()
Audio(sig_wav_h, rate=fs)
```

```python
sf.write("sig_wav_h.wav", sig_wav_h, fs)
```
